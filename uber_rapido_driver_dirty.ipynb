```python
## This code generates random date values for the onboarding column. 

import pandas as pd
import numpy as np

df['onboarding_date'] = pd.to_datetime(df['onboarding_date'], errors='coerce')

start = pd.Timestamp("2025-01-01")
end = pd.Timestamp("2026-08-11")

mask = df["onboarding_date"].isna()

df.loc[mask, "onboarding_date"] = pd.to_datetime(
    np.random.randint(
        start.value // 10**9,
        end.value // 10**9,
        size=mask.sum()
    ),
    unit="s"
)
df.sample(1000)
```
