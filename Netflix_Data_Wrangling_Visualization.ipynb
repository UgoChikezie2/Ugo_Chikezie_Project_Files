```python
## This code gives a visual representation of the values of the rating vs their release year using a Bar chart. 


import pandas as pd
import matplotlib.pyplot as plt

# Select recent years for clarity
df_plot = df[
    df['release_year'].between(2015, 2021)
].copy()

# Count titles by year and rating
rating_counts = pd.crosstab(
    df_plot['release_year'],
    df_plot['rating']
)

# Plot stacked bar chart
rating_counts.plot(
    kind='bar',
    stacked=True,
    figsize=(14, 7),
    colormap='tab20'
)

plt.title('Netflix Content Ratings by Release Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.legend(title='Rating', bbox_to_anchor=(1.05, 1),
           loc='upper left')
plt.tight_layout()
plt.show()
```

```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Convert date_added to datetime
df['date_added'] = pd.to_datetime(
    df['date_added'].astype('string').str.strip(),
    errors='coerce'
)

# Remove missing values
df_clean = df.dropna(subset=['country', 'date_added']).copy()

# Select the top 10 country entries
top_countries = df_clean['country'].value_counts().head(10).index

# Filter the dataset
df_top = df_clean[df_clean['country'].isin(top_countries)]

# Create scatterplot
plt.figure(figsize=(14, 7))

sns.scatterplot(
    data=df_top,
    x='country',
    y='date_added',
    hue='type',
    alpha=0.5,
    s=45
)

plt.title('Netflix Content Added by Country and Date', fontsize=16)
plt.xlabel('Country', fontsize=12)
plt.ylabel('Date Added', fontsize=12)

plt.xticks(rotation=45, ha='right')
plt.legend(title='Content Type')
plt.tight_layout()
plt.show()
```

```python

# Extract year
df_clean['year_added'] = df_clean['date_added'].dt.year

# Create a pivot table
heatmap_data = pd.crosstab(
    df_top['country'],
    df_top['date_added'].dt.year
)

plt.figure(figsize=(14, 7))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt='d',
    cmap='Blues'
)

plt.title('Netflix Titles Added by Country and Year')
plt.xlabel('Year Added')
plt.ylabel('Country')
plt.tight_layout()
plt.show()
```

```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Select movies only
movies = df[df['type'] == 'Movie'].copy()

# Extract numerical duration
movies['duration_minutes'] = pd.to_numeric(
    movies['duration'].str.extract(r'(\d+)')[0],
    errors='coerce'
)

# Remove missing values
movies = movies.dropna(subset=['duration_minutes'])

# Create boxplot
plt.figure(figsize=(12, 5))

sns.boxplot(
    data=movies,
    x='duration_minutes',
    color='skyblue'
)

plt.title('Distribution of Netflix Movie Durations', fontsize=16)
plt.xlabel('Duration (Minutes)', fontsize=12)

plt.tight_layout()
plt.show()
```

```python

Q1 = movies['duration_minutes'].quantile(0.25)
Q3 = movies['duration_minutes'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = movies[
    (movies['duration_minutes'] < lower_bound) |
    (movies['duration_minutes'] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Outliers:", len(outliers))

display(outliers[['title', 'duration', 'duration_minutes']].head(20))
```

```python
plt.figure(figsize=(12, 6))

sns.histplot(
    data=movies,
    x='duration_minutes',
    bins=30,
    kde=True,
    color='steelblue'
)

plt.title('Distribution of Netflix Movie Durations')
plt.xlabel('Duration (Minutes)')
plt.ylabel('Number of Movies')
plt.tight_layout()
plt.show()
```
