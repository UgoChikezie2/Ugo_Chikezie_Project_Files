```python
import pandas as pd

#import matplotlib.pyplot as plt

df = pd.read_csv("restaurant_sales_data.csv")
df.head(1000)
```

```python
## This converts the data type from object to string.
df['Item'] = df['Item'].astype("string")
print(df['Item'].dtype)
```

```python
df['Item'] = df['Item'].str.capitalize()
df['Item'] = df['Item'].fillna('Mnago Juice')
```

```python
df['Item'].unique()
```

```python
df['Price'].unique()
```

```python
## This converts the data type from object to string.
df['Price'] = df['Price'].astype("string")
print(df['Price'].dtype)
```

```python
df['Price'] = df['Price'].str.capitalize()
df['Price'] = df['Price'].fillna('11.5')
```

```python
df['Price'].unique()
```

```python
## This converts the data type from object to string.
df['Quantity'] = df['Quantity'].astype("string")
print(df['Quantity'].dtype)
```

```python
#df['Quantity'] = df['Quantity'].str.capitalize()
df['Quantity'] = df['Quantity'].fillna('9.5')
```

```python
df['Quantity'].unique()
```

```python
df['Order Total'].unique()
```

```python
df['Order Total'] = df['Order Total'].fillna('65')
```

```python
df['Payment Method'].unique()
```

```python
df['Order Total'] = df['Order Total'].fillna('Debit_Card')
```

```python
## This converts the data type from object to string.
df['Payment Method'] = df['Payment Method'].astype("string")
print(df['Payment Method'].dtype)
```

```python
df['Payment Method'].unique()
```

```python
df['Payment Method'] = df['Payment Method'].fillna('Debit_Card')
```

```python
df.info()
```

```python
df.sample(1000)
```

```python
print(df.isna().sum())
```

```python
## This code gives a visual representation of the values of the Items vs their Payment methods using a Bar chart. 

import matplotlib.pyplot as plt
%matplotlib inline

plt.bar(df['Payment Method'], df["Item"])

plt.title("Payment Method against Item")
plt.xlabel("Payment Method")
plt.ylabel("Item")

plt.show()
```

```python
## This code gives a visual representation of the values of the Items vs their Categories using a Bar chart. 

import matplotlib.pyplot as plt
%matplotlib inline

plt.bar(df['Category'], df["Item"])

plt.title("Category against Item")
plt.xlabel("Category")
plt.ylabel("Item")

plt.show()
```

```python
## This code gives a visual representation of the values of the Price using a Histogram. 

import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(data=df, x='Price')
plt.show()
```

```python
## This code gives a visual representation of the distribution of Items and their Prices using a Histogram. 


import matplotlib.pyplot as plt

plt.hist(df["Item"], bins=20, alpha=0.5, label="Item")
plt.hist(df["Price"], bins=20, alpha=0.5, label="Price")

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Distribution of Item and Price")
plt.legend()
plt.show()
```

```python
## This code gives a visual representation of the values of Items, Quantity, Price, and Payment Method using a Pie chart. 


import matplotlib.pyplot as plt

# Sample data
labels = ['Item', 'Quantity', 'Price', 'Payment Method']
sizes = [45, 35, 25, 10]

# Plot
plt.pie(sizes, labels=labels)

# Show chart
plt.show()

```
